```python
def modify_arguments(immutable_arg, mutable_arg):
    """
    Demonstrates the behavior of mutable and immutable arguments in Python.
    """
    print(f"Initial immutable_arg: {immutable_arg}, id: {id(immutable_arg)}")
    print(f"Initial mutable_arg: {mutable_arg}, id: {id(mutable_arg)}")

    immutable_arg = immutable_arg + 1  # Modifying immutable argument
    mutable_arg.append(4)  # Modifying mutable argument
    mutable_arg = [5, 6]  # Reassigning mutable argument

    print(f"Modified immutable_arg: {immutable_arg}, id: {id(immutable_arg)}")
    print(f"Modified mutable_arg: {mutable_arg}, id: {id(mutable_arg)}")


# Example usage
immutable_value = 10
mutable_list = [1, 2, 3]

modify_arguments(immutable_value, mutable_list)

print(f"Outside function - immutable_value: {immutable_value}")
print(f"Outside function - mutable_list: {mutable_list}")
```

```python
def basic_arithmetic(num1, num2):
    """Performs basic arithmetic operations on two numbers."""
    print(f"Addition: {num1 + num2}")
    print(f"Subtraction: {num1 - num2}")
    print(f"Multiplication: {num1 * num2}")
    if num2 != 0:  # Avoid division by zero
        print(f"Division: {num1 / num2}")
    else:
        print("Cannot divide by zero.")

# Get user input
try:
    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))
    basic_arithmetic(num1, num2)

except ValueError:
    print("Invalid input. Please enter numbers.")
```

```python
def demonstrate_logical_operators():
    """demonstrate the use of logical operators."""

    # Get input from the user
    x = int(input("Enter the first number:"))
    y = int(input("Enter the second number:"))

    # performs logical operations
    print("x and y:", x and y)  # True if both x and y are true
    print("x or y:", x or y)   #True if either x or y is true
    print("not x:", not x)    # True if x is False


    # call the function to execute the program
    demonstrate_logical_operators()
```

```python
my_list = ['1', '2', '3', '4.0', '5.5', 'hello']

int_list = []
float_list = []

for item in my_list:
    if item.isdigit():
        int_list.append(int(item))
    try:
        float_list.append(float(item))
    except ValueError:
        print(f"Cannot convert '{item}' to float.")

print("Integer list:", int_list)
print("Float list:", float_list)
```

```python
def check_number(num):
  """Checks if a number is positive, negative, or zero."""
  if num > 0:
    print("The number is positive.")
  elif num < 0:
    print("The number is negative.")
  else:
    print("The number is zero.")

# Example Usage
num = 5
check_number(num)  # Output: The number is positive.

num = -2
check_number(num) # Output: The number is negative.

num = 0
check_number(num) # Output: The number is zero.
```

```python
def reverse_string(s):
    reversed_s = ""
    length = len(s) - 1
    while length >= 0:
        reversed_s += s[length]
        length -= 1
    return reversed_s

input_string = "Hello, World!"
reversed_string = reverse_string(input_string)
print(f"Original string: {input_string}")
print(f"Reversed string: {reversed_string}")
```

```python
def factorial_while_loop(number):
  """
    Calculate the factorial of a number using a while loop.

    Args:
      number: An integer for which the factorial is to be calculated.

    Returns:
      The factorial of the number, or 1 if the number is 0.
      Returns a message if the number is negative.
  """
  if number < 0:
    return "Factorial is not defined for negative numbers."
  elif number == 0:
    return 1
  else:
    factorial = 1
    while number > 1:
      factorial *= number
      number -= 1
    return factorial

if __name__ == "__main__":
  num = int(input("Enter a non-negative integer: "))
  result = factorial_while_loop(num)
  print(f"The factorial of {num} is {result}")
```
